```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('image.jpeg')

img = cv2.resize(img, (300,300))
rgb = img[:,:,::-1]

plt.subplot(1,2,1)
plt.imshow(rgb)
print(rgb.shape)
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.subplot(1,2,2)
plt.imshow(rgb)
plt.axis('off')
print(rgb.shape)
```

```python
r = rgb[:,:, 0]
g = rgb[:,:, 1]
b = rgb[:,:, 2]

grey = (0.299 * r + .587* g + 0.114 * b).astype(np.uint8) 
plt.figure(figsize=(12,5))
plt.subplot(1,4,1)
plt.imshow(grey, cmap='grey')
plt.subplot(1,4,2)
plt.imshow(r, cmap='Reds')
plt.subplot(1,4,3)
plt.imshow(g, cmap='Greens')
plt.subplot(1,4,4)
plt.imshow(b, cmap='Blues')
print("Gray color shape = " , grey.shape)
```

```python
_, binary = cv2.threshold(grey, 127,255, cv2.THRESH_BINARY)
plt.subplot(1,2,1)
plt.imshow(binary , cmap='grey')

mannually = np.zeros_like(grey)

mannually[grey > 127] = 255

plt.subplot(1,2,2)
plt.imshow(mannually , cmap='grey')

```
