```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('image.jpeg')
img = img[:,:,::-1]

r = img[:,:,0]
g = img[:,:,1]
b = img[:,:,2]

rgb = np.zeros(img.shape, dtype=np.uint8)

rgb[:,:,0] = r
rgb[:,:,1] = g
rgb[:,:,2] = b

# rgb = np.stack(r,g,b)

gray = (r * .299 + g * .587 + b * .114).astype(np.uint8)

plt.figure(figsize=(12,3))
plt.subplot(1,3,1)
plt.imshow(img)
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(gray, cmap='gray')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(rgb)
plt.title("Restored image")
plt.axis('off')
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread('image.jpg', cv2.IMREAD_GRAYSCALE)


# 1. Spatial Resolution
sizes = [256, 128, 64, 32]

plt.figure(figsize=(10, 8))

for i, size in enumerate(sizes):
    small = cv2.resize(img, (size, size))
    result = cv2.resize(small, (img.shape[1], img.shape[0]))

    plt.subplot(2, 2, i + 1)
    plt.imshow(result, cmap='gray')
    plt.title(f'Spatial Resolution: {size} x {size}')
    plt.axis('off')

plt.tight_layout()
plt.show()

# 2. Intensity Resolution
levels = [256, 16, 4, 2]

plt.figure(figsize=(10, 8))

for i, L in enumerate(levels):
    step = 256 // L
    result = (img // step) * step

    plt.subplot(2, 2, i + 1)
    plt.imshow(result, cmap='gray', vmin=0, vmax=255)
    plt.title(f'Intensity Levels: {L}')
    plt.axis('off')

plt.tight_layout()
plt.show()
```
