## Implement a python program to implement image negative and  thresholding operation


```python
from pathlib import Path

import cv2
import matplotlib.pyplot as plt


gray = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)
```

```python

# Image negative: s = 255 - r for an 8-bit grayscale image.
negative = 255 - gray

# Binary thresholding: pixels above the threshold become white; the rest become black.
threshold_value = 127
_, thresholded = cv2.threshold(gray, threshold_value, 255, cv2.THRESH_BINARY)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(gray, cmap="gray", vmin=0, vmax=255)
axes[0].set_title("Original grayscale")
axes[1].imshow(negative, cmap="gray", vmin=0, vmax=255)
axes[1].set_title("Image negative")
axes[2].imshow(thresholded, cmap="gray", vmin=0, vmax=255)
axes[2].set_title(f"Binary threshold (T={threshold_value})")

for axis in axes:
    axis.axis("off")

plt.tight_layout()
plt.show()
```
