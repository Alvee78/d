11. Write a python program to apply log and power-law (gamma) transformations.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread('image.jpeg')

# Convert BGR to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert RGB to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
```

```python

# --------------------------------
# 1. LOG TRANSFORMATION
# --------------------------------

c = 255 / np.log(1 + np.max(gray))

log_image = c * np.log(1 + gray)

log_image = np.uint8(log_image)


# --------------------------------
# 2. POWER-LAW (GAMMA) TRANSFORMATION
# --------------------------------

gamma = 0.5
c = 1

power_image = c * np.power(gray / 255.0, gamma) * 255

power_image = np.uint8(power_image)


# --------------------------------
# 3. DISPLAY RESULTS
# --------------------------------

plt.figure(figsize=(12, 8))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(log_image, cmap='gray')
plt.title('Log Transformation')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(power_image, cmap='gray')
plt.title(f'Gamma Transformation (γ={gamma})')
plt.axis('off')

plt.tight_layout()
plt.show()
```
