## Write a python program to grey lelvel slicing with and without background


```python
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np


image_path = Path("image.jpeg")
if not image_path.is_file():
    raise FileNotFoundError(f"Image not found: {image_path.resolve()}")

gray = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)
if gray is None:
    raise ValueError(f"Unable to read image: {image_path.resolve()}")

low, high = 100, 200
highlight = 255
selected = (gray >= low) & (gray <= high)

# Without background preservation: keep only the selected gray-level range.
sliced_without_background = np.where(selected, highlight, 0).astype(np.uint8)

# With background preservation: retain other pixels from the original image.
sliced_with_background = gray.copy()
sliced_with_background[selected] = highlight

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(gray, cmap="gray", vmin=0, vmax=255)
axes[0].set_title("Original grayscale")
axes[1].imshow(sliced_without_background, cmap="gray", vmin=0, vmax=255)
axes[1].set_title(f"Slicing without background ({low}-{high})")
axes[2].imshow(sliced_with_background, cmap="gray", vmin=0, vmax=255)
axes[2].set_title(f"Slicing with background ({low}-{high})")

for axis in axes:
    axis.axis("off")

plt.tight_layout()
plt.show()
```
