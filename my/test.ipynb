```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread('image.jpeg')
img = cv.resize(img, (128,128))
img = img[:,:,::-1]
r = img[:,:,0]
g = img[:,:,1]
b = img[:,:,2]
rgb = np.stack([r,g,b], axis=2)

rgb = np.zeros_like(img, dtype=(np.uint8))
rgb[:,:,0] = r
rgb[:,:,1] = g
rgb[:,:, 2] = b
gray = (r * .299 + g * .587 + b*.114).astype(np.uint8)

gray = cv.cvtColor(rgb, cv.COLOR_RGB2GRAY)
img_bin = np.zeros_like(gray, dtype=(np.uint8))
img_bin[gray>127] = 255

plt.imshow(img_bin, cmap='gray')
```

```python
#spatial and intensity level..

# small = gray[::8,::8]

# restore = np.repeat(
#     (np.repeat(small,8,axis=0)),8,axis=1
# )

small = cv.resize(gray, (64,64))
restore = cv.resize(gray, gray.shape[::-1])

bit4 = gray // 64 * 84

bitres = bit4 // 84 * 64



plt.subplot(2,3,1)
plt.imshow(gray,cmap='gray')
plt.title(gray.shape)
plt.subplot(2,3,2)
plt.imshow(small,cmap='gray')
plt.subplot(2,3,3)
plt.imshow(restore,cmap='gray')
plt.title(f"restored = {restore.shape}")
plt.subplot(2,3,4)
plt.imshow(gray,cmap='gray')
plt.subplot(2,3,5)
plt.imshow(bit4,cmap='gray')
plt.subplot(2,3,6)
plt.imshow(bitres,cmap='gray')
```
