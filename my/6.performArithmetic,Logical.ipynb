```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Load base image and create a geometric template mask
img1 = cv2.imread('image.jpeg')
img1 = cv2.resize(img1, (400, 400)) # Ensure uniform sizing for operations
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)

# Create a second black canvas with a solid white circle in the middle
mask = np.zeros((400, 400), dtype=np.uint8)
cv2.circle(mask, (200, 200), 100, 255, -1)
# --- Arithmetic Operations ---
img_add = cv2.add(gray1, mask)
img_sub = cv2.subtract(gray1, mask)

# --- Logical Operations ---
bit_and = cv2.bitwise_and(gray1, mask)
bit_or = cv2.bitwise_or(gray1, mask)
bit_not = cv2.bitwise_not(gray1)

# Plot everything out
fig, axs = plt.subplots(2, 3, figsize=(15, 10))
axs[0, 0].imshow(gray1, cmap='gray'); axs[0, 0].set_title("Original Gray"); axs[0, 0].axis('off')
axs[0, 1].imshow(img_add, cmap='gray'); axs[0, 1].set_title("Arithmetic ADD"); axs[0, 1].axis('off')
axs[0, 2].imshow(img_sub, cmap='gray'); axs[0, 2].set_title("Arithmetic SUBTRACT"); axs[0, 2].axis('off')

axs[1, 0].imshow(bit_and, cmap='gray'); axs[1, 0].set_title("Logical AND"); axs[1, 0].axis('off')
axs[1, 1].imshow(bit_or, cmap='gray'); axs[1, 1].set_title("Logical OR"); axs[1, 1].axis('off')
axs[1, 2].imshow(bit_not, cmap='gray'); axs[1, 2].set_title("Logical NOT"); axs[1, 2].axis('off')
plt.show()
```
